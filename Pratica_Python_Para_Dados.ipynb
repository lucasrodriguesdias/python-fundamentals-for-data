Para o diesel, se a quantidade comprada for até 15 litros, o desconto será de 3% por litro. Caso contrário, será de 5% por litro. O preço do litro de diesel é R\$ 2,00 e o preço do litro de etanol é R\$ 1,70. Escreva um programa que leia a quantidade de litros vendidos e o tipo de combustível (E para etanol e D para diesel) e calcule o valor a ser pago pelo cliente. Tenha em mente algumas dicas:

- O do valor do desconto será a multiplicação entre preço do litro, quantidade de litros e o valor do desconto.
- O valor a ser pago por um cliente será o resultado da multiplicação do preço do litro pela quantidade de litros menos o valor de desconto resultante do cálculo.


````python
litroEtanol = 1.7
litroDiesel = 2

combustivel = input('Qual combustível deseja colocar (etanol ou diesel)? ')
qtdLitros = float(input('Quantos litros deseja? '))

if combustivel == 'etanol':
  subtotal = float(qtdLitros * litroEtanol)
else:
  subtotal = float(qtdLitros * litroDiesel)

## Cálculo desconto
if combustivel == 'etanol' and qtdLitros <= 15:
  descontoEtanol = litroEtanol * qtdLitros * 0.02

elif combustivel == 'etanol' and qtdLitros > 15:
  descontoEtanol = litroEtanol * qtdLitros * 0.04

elif combustivel == 'diesel' and qtdLitros <= 15:
  descontoDiesel = litroDiesel * qtdLitros * 0.03

else:
  descontoDiesel = litroDiesel * qtdLitros * 0.05

## Valor total a ser pago

print('\n###Cobrança Final###\n')
print('Combustível escolhido = ', combustivel)
print('Quantidade em litros = ', qtdLitros)
print (f'Subtotal = R$ {subtotal}')

if combustivel == 'etanol':
  print('Desconto aplicado = R$ ', descontoEtanol)
else:
  print('Desconto aplicado = R$ ', descontoDiesel)

if combustivel == 'etanol':
  print('\n###############')
  print('Valor total a ser pago = R$ ', subtotal - descontoEtanol)
else:
  print('\n###############')
  print('Valor total a ser pago = R$ ', subtotal - descontoDiesel)
````

#### Questão 2
Escreva um programa para calcular quantos dias levará para a colônia de uma bactéria A ultrapassar ou igualar a colônia de uma bactéria B, com base nas taxas de crescimento de 3% e 1,5% respectivamente. Considere que a colônia A inicia com 4 elementos e a colônia B com 10 elementos.

````python
coloniaA = 4
coloniaB = 10
cresceA = 0.03
cresceB = 0.015
tempo = 0 #registro em dias conforme o enunciado

while coloniaA < coloniaB:
  coloniaA = coloniaA + (coloniaA*cresceA)
  coloniaB = coloniaB + (coloniaB*cresceB)

  tempo += 1

print(f'O tempo para a colônia A igualar/ultrapassar a colônia B foi de {tempo} dias.\n')
print(f'Colônia A = {coloniaA}')
print(f'Colônia B = {coloniaB}\n')
````

````python
amostraDiaria = [1.2, 2.1, 3.3, 5.0, 7.8, 11.3, 16.6, 25.1, 37.8, 56.9]
crescimento = []

for i in range(1, len(amostraDiaria)):
  amostra_atual = amostraDiaria[i]
  amostra_passada = amostraDiaria[i-1]

  porcentagemCrescimento = 100 * (amostra_atual - amostra_passada) / (amostra_passada)
  crescimento.append(porcentagemCrescimento)

print(f'As porcentagens diárias de cresimento da colônia foram: {crescimento}')
````

````python
setorIdade = {'Setor A': [22, 26, 30, 30, 35, 38, 40, 56, 57, 65],
              'Setor B': [22, 24, 26, 33, 41, 49, 50, 54, 60, 64],
              'Setor C': [23, 26, 26, 29, 34, 35, 36, 41, 52, 56],
              'Setor D': [19, 20, 25, 27, 34, 39, 42, 44, 50, 65]}

setorMedia = {}


for setor, listaIdade in setorIdade.items():
  mediaSetor = sum(listaIdade) / len(listaIdade)
  setorMedia[setor] = mediaSetor

somaIdadeGeral = 0
qtdFunciorarioSetor = 0

for listaIdade in setorIdade.values():
  somaIdadeGeral += sum(listaIdade)
  qtdFunciorarioSetor += len(listaIdade)

mediaGeral = somaIdadeGeral / qtdFunciorarioSetor

acimaMedia = 0

for listaIdade in setorIdade.values():
  for idade in listaIdade:
    if idade > mediaGeral:
      acimaMedia += 1

print(f'A média de idade de cada setor é:')
print(f'{setorMedia}\n')
print(f'A idade média geral da empresa é de {mediaGeral:.2f} anos.\n')
print(f'{acimaMedia} funcionários possuem idade acima da média.')
````
